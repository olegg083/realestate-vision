```python
import os
import shutil
import random
from pathlib import Path
```

```python
random.seed(42)

RAW_DATA_DIR = Path('../data/raw/indoorCVPR_09/Images') 
PROCESSED_DATA_DIR = Path('../data/processed')

TARGET_CLASSES = ['bathroom', 'bedroom', 'dining_room', 'kitchen', 'livingroom']
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
```

```python
def prepare_dataset():
    if PROCESSED_DATA_DIR.exists():
        shutil.rmtree(PROCESSED_DATA_DIR)

    train_dir = PROCESSED_DATA_DIR / 'train'
    val_dir = PROCESSED_DATA_DIR / 'val'
    test_dir = PROCESSED_DATA_DIR / 'test'
    
    
    for class_name in TARGET_CLASSES:
        os.makedirs(train_dir / class_name, exist_ok=True)
        os.makedirs(val_dir / class_name, exist_ok=True)
        os.makedirs(test_dir / class_name, exist_ok=True)
        
        class_dir = RAW_DATA_DIR / class_name
        images = sorted(class_dir.glob('*.jpg'))
        random.shuffle(images)
        
        n_total = len(images)
        train_end = int(n_total * TRAIN_RATIO)
        val_end = train_end + int(n_total * VAL_RATIO)
        
        train_imgs = images[:train_end]
        val_imgs = images[train_end:val_end]
        test_imgs = images[val_end:]
        
        for img in train_imgs: shutil.copy(img, train_dir / class_name / img.name)
        for img in val_imgs: shutil.copy(img, val_dir / class_name / img.name)
        for img in test_imgs: shutil.copy(img, test_dir / class_name / img.name)
            
        print(f"{class_name:12s} | Train: {len(train_imgs):3d} | Val: {len(val_imgs):3d} | Test: {len(test_imgs):3d}")

prepare_dataset()
print("Датасет подготовлен")
```
