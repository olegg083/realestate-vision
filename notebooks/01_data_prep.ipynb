```python
import os
import shutil
import random
from pathlib import Path
```

```python
random.seed(42)

RAW_DATA_DIR = Path('../data/raw/indoorCVPR_09/Images') 
PROCESSED_DATA_DIR = Path('../data/processed')

TARGET_CLASSES = ['bathroom', 'bedroom', 'dining_room', 'kitchen', 'livingroom']
SPLIT_RATIO = 0.8 # 80% на обучение, 20% на валидацию
```

```python
def prepare_dataset():
    train_dir = PROCESSED_DATA_DIR / 'train'
    val_dir = PROCESSED_DATA_DIR / 'val'
    
    for class_name in TARGET_CLASSES:
        os.makedirs(train_dir / class_name, exist_ok=True)
        os.makedirs(val_dir / class_name, exist_ok=True)
        
        class_dir = RAW_DATA_DIR / class_name
        if not class_dir.exists():
            print(f"Папка {class_dir} не найдена")
            continue
            
        images = list(class_dir.glob('*.jpg'))
        random.shuffle(images)
    
        split_idx = int(len(images) * SPLIT_RATIO)
        train_images = images[:split_idx]
        val_images = images[split_idx:]
        
        for img_path in train_images:
            shutil.copy(img_path, train_dir / class_name / img_path.name)
            
        for img_path in val_images:
            shutil.copy(img_path, val_dir / class_name / img_path.name)
            
        print(f"Класс {class_name}: {len(train_images)} train, {len(val_images)} val")

prepare_dataset()
print("Датасет подготовлен")
```
